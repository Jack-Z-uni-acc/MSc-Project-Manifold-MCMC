```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 1000 # Number of iterations to run RWMH for each parameter value
n = 100 # Number of dimensions to test over
seed = 0 # Seed for random number generator
X0 = 0 # Initial value for all tests
```

```python
# Log-acceptance probability
def log_r(x, y):
  return 0.5 * (np.linalg.norm(x)**2 - np.linalg.norm(y)**2)
```

```python
# RWMH step
def step(x, beta, xi, U):

  px = x + beta * xi
  acc = min(0,log_r(x, px))

  if acc < U:
    y = x
    success = 0

  else:
    y = px
    success = 1

  return y, acc, success
```

```python
# Scaling with fixed beta = 1
beta = 1
E_log_r = np.zeros(n)
E_alpha = np.zeros(n)

for i in range(n):

  np.random.seed(seed + i)
  xis = np.random.normal(size=(m,i))
  Us = np.log(np.random.uniform(0,1,m))

  rval = 0
  sval = 0
  x = X0

  for j in range(m):

    x, r, s = step(x, beta, xis[j], Us[j])
    rval += r
    sval += s

  rval /= m
  sval /= m
  E_log_r[i] = rval
  E_alpha[i] = sval
```

```python
fig, ax = plt.subplots()
xvals = np.arange(0,n)
ax.plot(xvals,E_log_r,label='Log-acceptance probability')
ax.plot(xvals,-xvals/2,'--',label='y = -0.5*N')
ax.set_title('Log-acceptance probability of RWMH at fixed proposal variance 1.')
ax.set_ylabel('Log-acceptance probability')
ax.set_xlabel('Dimension N')
ax.grid()
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(xvals[:50],E_alpha[:50],label='Empirical expected acceptance probability')
ax.plot(xvals[:50], np.exp(-0.2*xvals[:50]),'--',label='y = exp(-0.2*N)')
ax.set_title('Empirical expected acceptance probability of RWMH at fixed proposal variance 1.')
ax.set_ylabel('Acceptance probability')
ax.set_xlabel('Dimension N')
ax.legend()
ax.grid()
plt.show()
```

```python
# Scaling with beta = N**(-0.5)

E_log_r_2 = np.zeros(n)
E_alpha_2 = np.zeros(n)

for i in range(n):

  np.random.seed(seed + i)
  xis = np.random.normal(size=(m,i))
  Us = np.log(np.random.uniform(0,1,m))

  rval = 0
  sval = 0
  x = X0

  for j in range(m):

    x, r, s = step(x, n**(-0.5), xis[j], Us[j])
    rval += r
    sval += s

  rval /= m
  sval /= m
  E_log_r_2[i] = rval
  E_alpha_2[i] = sval
```

```python
fig, ax = plt.subplots()
xvals = np.arange(0,n)
ax.plot(xvals,E_log_r_2,label='Log-acceptance probability')
ax.set_title('Log-acceptance probability of RWMH at proposal variance N^(-0.5).')
ax.set_ylabel('Log-acceptance probability')
ax.set_xlabel('Dimension N')
ax.set_ylim(-5,5)
ax.grid()
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(xvals,E_alpha_2,label='Empirical expected acceptance probability')
ax.set_title('Empirical expected acceptance probability of RWMH at proposal variance N^(-0.5).')
ax.set_ylabel('Acceptance probability')
ax.set_xlabel('Dimension N')
ax.grid()
plt.show()
```
